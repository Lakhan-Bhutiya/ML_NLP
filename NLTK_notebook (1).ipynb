# 📘 NLP Analysis Using NLTK, Gutenberg, Brown & Vectorization

This notebook demonstrates:
- Downloading NLTK corpora  
- Tokenization  
- Cleaning text using Regex + Stopwords  
- Word frequency analysis  
- Bag of Words  
- Binary Vectorization  
- TF  
- TF-IDF  
- Word2Vec  



```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import gutenberg, brown, stopwords

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("gutenberg")
nltk.download("brown")
```

###  Observation
- `punkt` → tokenizer model for splitting text  
- `stopwords` → list of 179 common English words  
- `gutenberg` & `brown` → NLP corpora  
These downloads are necessary for tokenization, cleaning, and analysis.


```python
text_g = gutenberg.raw("austen-emma.txt")
text_g[:300]
```

###  Observation
We loaded **Emma** by Jane Austen.  
This is a large novel (160k+ words), useful for analyzing real English literature.


```python
words_g = word_tokenize(text_g.lower())
words_g[:20]
```

### Observation
- Tokenization splits text into individual units  
- Includes punctuation → will be removed later  
- Lowercasing ensures uniform text


```python
stop = set(stopwords.words("english"))
clean_g = [w for w in words_g if w.isalpha() and w not in stop]
clean_g[:20]
```

###  Observation
We removed:
- Punctuation (`isalpha()`)  
- 179 stopwords (the, is, and, etc.)  

Clean text now contains meaningful words — required for TF-IDF & vectors.


```python
df_g = pd.DataFrame(clean_g, columns=["word"])
df_g.head()
```

###  Observation
Creating a DataFrame helps:
- frequency counting  
- visualization  
- exporting cleaned tokens  


```python
from nltk.probability import FreqDist

freq_g = FreqDist(clean_g)
freq_g.most_common(10)
```

### 📌 Observation
Most frequent words reflect Austen's writing style.  
In *Emma*, words like **said, miss, emma, could** dominate.


```python
text_b = " ".join(brown.words()[:2000])
text_b[:300]
```

### 📌 Observation
Brown corpus contains:
- News  
- Editorial  
- Fiction  
- Government writing  

Much more formal and varied than Gutenberg literature.


```python
words_b = word_tokenize(text_b.lower())
clean_b = [w for w in words_b if w.isalpha() and w not in stop]

clean_b[:20]
```

### 📌 Observation
Cleaning process same as Gutenberg.  
Brown vocabulary is more modern and factual.


```python
corpus = [
    " ".join(clean_g[:3000]),
    " ".join(clean_b[:3000])
]

df = pd.DataFrame({"text": corpus})
df
```

### 📌 Observation
We now have **two documents**:
1. Gutenberg (literature)
2. Brown (formal modern English)

Perfect for TF-IDF comparison.


```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

bow = CountVectorizer()
bow_matrix = bow.fit_transform(df["text"])

df_bow = pd.DataFrame(bow_matrix.toarray(), columns=bow.get_feature_names_out())
df_bow.head()
```

### 📌 Observation
Bag of Words counts frequency of each word.
Higher number → word appears more often.
Sparse matrix because many words do not appear in both corpora.


```python
binary = CountVectorizer(binary=True)
binary_matrix = binary.fit_transform(df["text"])

df_binary = pd.DataFrame(binary_matrix.toarray(), columns=binary.get_feature_names_out())
df_binary.head()
```

### 📌 Observation
Binary vectorizer uses:
- 1 → word present
- 0 → word absent  
Does NOT care about frequency.
Useful for classification tasks.


```python
tf = TfidfVectorizer(use_idf=False, norm=None)
tf_matrix = tf.fit_transform(df["text"])

df_tf = pd.DataFrame(tf_matrix.toarray(), columns=tf.get_feature_names_out())
df_tf.head()
```

### 📌 Observation
TF = word_count / total_words_in_document  
This normalizes frequencies when documents are different lengths.


```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df["text"])

df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf.get_feature_names_out())
df_tfidf.head()
```

### 📌 Observation
TF-IDF gives:
- LOW score → common words  
- HIGH score → rare but important words  
Better for text mining, document similarity, NLP tasks.


```python
from gensim.models import Word2Vec

tokenized = [doc.split() for doc in df["text"]]

model = Word2Vec(
    sentences=tokenized,
    vector_size=20,
    window=5,
    min_count=1,
    workers=4
)
```

### 📌 Observation
Word2Vec creates **dense semantic vectors**.
Words with similar meaning get closer vector positions.
Unlike TF-IDF, Word2Vec understands context.


```python
model.wv["emma"]  # try any word present in corpus
```

### 📌 Observation
This outputs a 20-dimensional vector.
Numbers represent semantic meaning learned from context.
High-level NLP models use embeddings like these.
